```python
reference_date = Sampledf['Order Date'].max() + pd.Timedelta(days=1)

print("Reference Date:", reference_date)
```

```python
customer_recency = (
    Sampledf.groupby('Customer ID')['Order Date']
    .max()
    .reset_index()
)

customer_recency['Recency'] = (
    reference_date - customer_recency['Order Date']
).dt.days

customer_recency.head()
```

frequency calculation

```python
customer_frequency = (
    Sampledf.groupby('Customer ID')['Order ID']
    .nunique()
    .reset_index()
)

customer_frequency.rename(
    columns={'Order ID': 'Frequency'},
    inplace=True
)

customer_frequency.head()
```

this is monetary calculation

```python
customer_monetary = (
    Sampledf.groupby('Customer ID')['Sales']
    .sum()
    .reset_index()
)

customer_monetary.rename(
    columns={'Sales': 'Monetary'},
    inplace=True
)

customer_monetary.head()
```

```python
rfm = customer_recency[['Customer ID', 'Recency']].merge(
    customer_frequency,
    on='Customer ID'
).merge(
    customer_monetary,
    on='Customer ID'
)

rfm.head(10)
```

here is the RFM scoring done.

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['RFM_Score'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)

rfm.head(10)
```

Customer Segementation

```python
def customer_segment(row):

    r, f, m = row['R_Score'], row['F_Score'], row['M_Score']

    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'

    elif r >= 3 and f >= 4 and m >= 3:
        return 'Loyal Customers'

    elif r >= 4 and f <= 3 and m >= 2:
        return 'Potential Loyalists'

    elif r == 5 and f <= 2:
        return 'New Customers'

    elif r <= 2 and f >= 4 and m >= 3:
        return 'At Risk'

    elif r <= 2 and f <= 3 and m >= 3:
        return 'Need Attention'

    else:
        return 'Lost Customers'


rfm['Customer_Segment'] = rfm.apply(customer_segment, axis=1)

rfm.head(10)
```

```python
segment_counts = (
    rfm['Customer_Segment']
    .value_counts()
    .reset_index()
)

segment_counts.columns = ['Customer_Segment', 'Customers']

segment_counts
```

```python
customer_segment_value = (
    Sampledf.merge(
        rfm[['Customer ID', 'Customer_Segment']],
        on='Customer ID',
        how='left'
    )
    .groupby('Customer_Segment')
    .agg(
        Sales=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
        Orders=('Order ID', 'nunique'),
        Quantity=('Quantity', 'sum')
    )
    .reset_index()
)

customer_segment_value['Profit Margin (%)'] = (
    customer_segment_value['Profit'] /
    customer_segment_value['Sales']
) * 100

customer_segment_value.sort_values('Sales', ascending=False)
```
